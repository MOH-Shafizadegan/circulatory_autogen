```python
import pandas as pd

csv_path = "cerebral_parameters_unfinished.csv"
df = pd.read_csv(csv_path)

# Find rows where variable_name starts with "u_ext" and value is "EMPTY_MUST_BE_FILLED"
mask = df["variable_name"].str.startswith("u_ext") & (df["value"].astype(str) == "EMPTY_MUST_BE_FILLED")

# Fill in 0 for value and "user_defined" for data_reference
df.loc[mask, "value"] = 0
df.loc[mask, "data_reference"] = "user_defined"

# Save the updated CSV
df.to_csv(csv_path, index=False)

print(f"Updated {mask.sum()} rows")
df[mask]
```
